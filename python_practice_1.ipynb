### 1. Reverse a string without slicing

```python
s = "hello"
rev = ""
for i in s:
    rev = i + rev
print(rev)
```

### 3. Count vowels

```python
s = "hello"
vowels = "aeiou"
count = 0
for ch in s:
    if ch in vowels:
        count += 1
print(count)
```

### 4. Character count

```python
s = "hello"
d = {}
for i in s:
    d[i] = d.get(i, 0) + 1
print(d)
```

### 5. Remove duplicates

### 24. Armstrong number

```python
n = 153
order = len(str(n))
print(sum(int(d)**order for d in str(n)) == n)
```

```python
n = 153

order = len(str(n))
total = 0

for i in str(n):
    total += int(i) ** order

if total == n:
    print("Armstrong")
else:
    print("Not Armstrong")
```

### 34. Max in list

```python
def find_max(lst):
    m = lst[0]
    for i in lst:
        if i > m:
            m = i
    return m
```

### 35. Recursive factorial

```python
def fact(n):
    if n==0:
        return 1
    return n*fact(n-1)
```

### 36. Recursive Fibonacci

```python
def fib(n):
    if n<=1:
        return n
    return fib(n-1)+fib(n-2)
```

### 37. Count even/odd

```python
def count_even_odd(lst):
    even=odd=0
    for i in lst:
        if i%2==0:
            even+=1
        else:
            odd+=1
    return even,odd
```
